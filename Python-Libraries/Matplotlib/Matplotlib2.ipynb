```python
# Histograms

# generate random data
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# generates 100 nums with mean 70 & std dev 10
scores = np.random.normal(loc=70, scale=10, size=100) 

plt.hist(scores, bins=5, color="skyblue", edgecolor="black")
plt.title("Distribution of exam scores")
plt.xlabel("Marks")
plt.ylabel("Number of students")
# print(scores)
```

```python
# custom bins
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]
# custom bins
bins = [0, 30, 70, 90, 100]
plt.hist(marks, bins=bins, color="lightblue", edgecolor="black")

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")

plt.grid(True)
```

```python
# axv line
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

# custom bins
bins = [0, 30, 70, 90, 100]
plt.hist(
    marks,
    bins=bins,
    edgecolor="black",
    color="lightblue"
)

# line at passing marks - 33
plt.axvline(
    x=33, 
    color="red", 
    linestyle="--",
    linewidth=2,
    label="Pass Mark (33)"
)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
plt.legend()
plt.grid(True)
```

```python
# Multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, bins=10, label="Fraud", color="red", alpha=0.5, edgecolor="black")
plt.hist(legit_transactions, bins=10, label="Legit", color="green", alpha=0.5, edgecolor="black")

plt.xlabel('Transaction Amount ($)')
plt.ylabel('Frequency')
plt.title('Transaction Amount Distribution: Legitimate vs Fraudulent')

plt.legend()
plt.grid(True)
```

```python
# Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.grid(True)
```

```python
# Multiple Datasets
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels=["Group1", "Group2"])
plt.show()
```

```python
# Horizontal Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False, showmeans=True, whis=2.0)
plt.grid(True)
```

```python
# Stack Plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

# Single day plot
# plt.pie([direct[0], organic[0], social[0]], labels=["direct", "organic", "social"], autopct="%1.1f%%")

# For multiple days
plt.stackplot(days, direct, organic, social, labels=["Direct", "Organic", "Social"])
plt.legend(loc='upper left')
plt.title("Website Traffic Over a Week")
plt.xlabel("Days")
plt.ylabel("Number of Visitors")
```
