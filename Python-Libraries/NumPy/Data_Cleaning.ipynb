```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("Store_data.json")
print(data)
print(type(data))
```

```python
# Clean & Structure the data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # Clean ratings - data consistency
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]

        user["rating"] = raw_rating

        # Handle missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # Deduplication
        if(user["name"].strip() in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
cleaned_data = clean_data(data)
```

```python
# Get meaningful insights from data
def get_insights(cleaned_data):
    # Avg rating
    tot_rating = 0
    for user in cleaned_data:
        tot_rating += float(user["rating"])

    avg_rating = tot_rating / len(cleaned_data)
    print(f"Avg rating = {avg_rating}")

    # Percentage of users with poor ratings
    poor_ratings = 0
    for user in cleaned_data:
        if(float(user["rating"]) < 3):
            poor_ratings += 1

    print(f"% of user with poor rating = {poor_ratings/len(cleaned_data) * 100}%")
```

```python
get_insights(cleaned_data)
```

```python
# Build recommendation feature
def get_recommendations(cleaned_data):
    recommendations = []

    for user in cleaned_data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendations(cleaned_data)
```
