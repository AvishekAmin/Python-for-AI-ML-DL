```python
# Web Scraping

import requests

URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
if res.status_code == 200:
    # print(res.text)
    # print(res.content)
    print(res.headers)

with open("Scraped_data/Data1.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
# BeautifulSoup
from bs4 import BeautifulSoup

with open("Scraped_data/Data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h3")
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)  # country name
    # print(h3.find_parent("div")["class"])
    # population = print(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))  # population values
    population = (
        h3.find_next("div")
            .select_one("span.country-population")
            .get_text(strip=True)
    )   # same result as previous

    all_countries.append([name, population])
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns=["Name", "Population"])
df
```

```python
df.to_csv("Cleaned_data/Data1.csv", index=False)
```
