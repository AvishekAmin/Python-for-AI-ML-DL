```python
import requests
from bs4 import BeautifulSoup
from pathlib import Path
```

```python
# Fetch data & store it
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No more valid quotes available!")
        break

    with open(f"Scraped_data/Quotes/Quote{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page {page_count}")

    page_count += 1
```

```python
# Extract useful info
page_count = 1
life_quotes = []

while True:
    file_path = Path(
        f"Scraped_data/Quotes/Quote{page_count}.html"
    )

    if not file_path.exists():
        break

    with open(file_path, "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")

    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = []
        
        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text(strip=True))

        if "life" in all_tags:
            text = q.select_one("span.text").get_text(strip=True)
            author = q.select_one("small.author").get_text(strip=True)
            life_quotes.append([text, author])

    page_count += 1
```

```python
print(f"Total quotes tagged 'life': {len(life_quotes)}")

for text, author in life_quotes:
    print(f"\nQuote: {text}")
    print(f"Author: {author}")
```
