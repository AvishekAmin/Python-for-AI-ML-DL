```python
import requests
from bs4 import BeautifulSoup
from pathlib import Path
```

```python
# Fetch data & store it
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No more valid quotes available!")
        break

    with open(f"Scraped_data/Quotes/Quote{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page {page_count}")

    page_count += 1
```
