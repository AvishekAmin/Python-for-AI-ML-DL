```python
import pandas as pd
```

```python
# Load data
df = pd.read_csv("Titanic_dataset.csv")
```

```python
# Display only columns: Name, Sex, Age, Fare, Survived
selected_columns = df[
    ["Name", "Sex", "Age", "Fare", "Survived"]
]

selected_columns
```

```python
# Select passengers who: Are female & Fare > 30
female_passengers = df[
    (df["Sex"] == "female") &
    (df["Age"] > 30)
]

female_passengers
```

```python
# Group by Pclass and compute: Survival rate, Average Fare and Average Age
class_stats = df.groupby("Pclass").agg(
    survival_rate=("Survived", "mean"),
    avg_fare=("Fare", "mean"),
    avg_age=("Age", "mean")
)

class_stats["survival_rate"] *= 100

class_stats
```
