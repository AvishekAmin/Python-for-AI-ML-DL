```python
import pandas as pd
```

```python
# Data cleaning
df = pd.read_csv("Raw_data.csv")

# Handling missing values
df.isnull()     # valid cells => False, missing cells => True
df.isnull().sum()       # missing values per column
df.dropna()     # deletes all rows that consists any missing value cell
df.fillna(0)    # fills all missing cells with 0

age_mean = df["age"].mean()
df["age"].fillna(age_mean)  # fills missing cells from age column with age_mean

cleaned_data = df.copy()    # copies the data in cleaned_data
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data

df.ffill()      # forward fill
df.bfill()      # backward fill
```

```python
# Handling duplicate values
df.duplicated()     # check whole rows one by one with each other

df["country"].duplicated()  # check only "country" column if same country (USA) appears duplicate later
df[["country", "gender"]].duplicated()      # # check "country" and "gender" column combination if same country (USA) & gender (Male) appears duplicate later

df2 = df.copy()
df.drop_duplicates(inplace=True)    # duplicates will be deleted in the actual dataframe
```

```python
# Data types
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64").copy()
df2.dtypes
```

```python
df["gender"].str.upper()        # All uppercase
df["gender"].str.lower()        # All lowercase
df["gender"].str.capitalize()   # First letter in capital, other in small
df["name"].str.split(" ")       # splits the full names using " "
df["country"].str.contains("US")# if country contains US => prints True, otherwise False
df["country"].str.contains("india", case=False) # ignores case sensitivity
```

```python
# Feature Engineering
df2 = df.copy()

# Apply
df2["tax"] = df2["income"].apply(lambda x: "20%" if x >= 60000 else "10%")

# Map
gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)

# Assign
df2 = df2.assign(new_income = df2["income"] * 1.1) 

# Replace
df2["country"] = df2["country"].replace("USA", "US")     # replaces old value (USA) with new value (US)

# Rename
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]
df2.rename(columns={"Income":"Salary"})

# Sort
df2.sort_values("Income")
df2.sort_values("Income", ascending=False)
sorted_df = df2.sort_values(["Income", "Age"])
sorted_df.sort_index()

# Reset
sorted_df.reset_index()
sorted_df.reset_index(drop=True)

# Ranking
sorted_df["Ranking"] = sorted_df["Income"].rank()
sorted_df["Ranking"] = sorted_df["Income"].rank(ascending=False, method="min")
sorted_df
```

```python
# Group by
df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].min()
df.groupby("gender")["income"].max()

# agg, aggregate functions are same
df.groupby("country")["income"].agg(["mean", "min", "max"])
df.groupby("country")["income"].aggregate(["mean", "min", "max"])

df.groupby("country")["income"].agg(avg_salary="mean", min_salary="min", max_salary="max")

df.groupby("country").agg({
    "income": "mean",
    "age": "mean"
})

df.groupby("country").agg(
    max_salary=("income", "max"),
    min_age=("age", "min")
)
```
