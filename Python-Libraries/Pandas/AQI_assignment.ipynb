```python
# Working with AQI dataset
import numpy as np
import pandas as pd
```

```python
# Reading the data from CSV  file
df = pd.read_csv("AQI_dataset.csv")

df.head(5)
```

```python
# Selecting the data

# columns
df[["city", "aqi"]]

# rows
df.loc[2]       # prints row data
df.loc[0:2]     # start idx : end idx (inclusive)

df.iloc[2]      # prints row data
df.iloc[0:2]    # start idx : end idx (non-inclusive)

# cells - (row, col)
df.loc[0, ["city", "aqi"]]
df.loc[0:2, ["city", "aqi"]]

df.iloc[0, [2, 11]]     # index: city => 2, aqi => 11
df.iloc[0:2, [2, 11]]

# cell
df.at[0, "city"]
df.iat[0, 2]        # 2 => city index
```

```python
# Filtering of data => returns a view of data
df[df["aqi"] > 100]     # columns with AQI > 100
df[(df["aqi"] > 100) & (df["temperature"] > 30)]    # columns with AQI > 100 & Temp > 30

df[df["aqi"] > 100][["city", "aqi"]]
```

```python
# Query method => returns a copy of data
aqi_val = 100
df.query("aqi > 100 & temperature > 30")
df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]]
```
