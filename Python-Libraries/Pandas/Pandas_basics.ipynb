```python
import pandas as pd
```

```python
# Series
s = pd.Series([6, 7, 8, 9, 10])
print(s)
print(type(s))

# index - based access
print(s[0])
print(s[3])

print(s.index)
```

```python
s = pd.Series([23, 24, 25, 26, "ABC"], index=["Adam", "Eve", "Bob", "Charlie", 30])
print(s)
print(type(s))

# label - based access
print(s["Adam"])
print(s["Charlie"])

print(s.index)
```

```python
# Vectorization
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])

print(s1 + s2)
```

```python
# DataFrame
info = {
    "Name": ["Avishek", "Rahul", "Tanmay"],
    "Age": [23, 24, 25],
    "GPA": [9.5, 8.6, 7.2]
}

df = pd.DataFrame(info)
print(df)
print(type(df))

print(df.index)
print(df.columns)
```

```python
df = pd.DataFrame([["Avishek", 23], ["Rahul", 24], ["Tanmay", 25]], columns=["Name", "Age"])

print(df)
```

```python
import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])

df = pd.DataFrame(np_arr, columns=["A", "B", "C"])
print(df)
```

```python
# Read CSV data
df = pd.read_csv("Employee_data.csv")
print(df)
print(type(df))
```

```python
# Read JSON data
df = pd.read_json("Employee_data.json")
print(df)
print(type(df))
```
