```python
import pandas as pd
```

```python
# Series
s = pd.Series([6, 7, 8, 9, 10])
print(s)
print(type(s))

# index - based access
print(s[0])
print(s[3])

print(s.index)
```

```python
s = pd.Series([23, 24, 25, 26, "ABC"], index=["Adam", "Eve", "Bob", "Charlie", 30])
print(s)
print(type(s))

# label - based access
print(s["Adam"])
print(s["Charlie"])

print(s.index)
```
